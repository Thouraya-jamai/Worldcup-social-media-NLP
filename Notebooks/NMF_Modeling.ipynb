```python
import pickle

nmf_model18 = pickle.load(
    open(r"C:\Projects\worldcup-social-media-nlp\models\worldcup2018_nmf.pkl", "rb")
)

vectorizer18 = pickle.load(
    open(r"C:\Projects\worldcup-social-media-nlp\models\worldcup2018_vectorizer.pkl", "rb")
)

nmf_model22 = pickle.load(
    open(r"C:\Projects\worldcup-social-media-nlp\models\worldcup2022_nmf.pkl", "rb")
)

vectorizer22 = pickle.load(
    open(r"C:\Projects\worldcup-social-media-nlp\models\worldcup2022_vectorizer.pkl", "rb")
)
```

```python
#2018 World Cup
feature_names = vectorizer18.get_feature_names_out()

for topic_idx, topic in enumerate(nmf_model18.components_[:10]):

    top_words = [
        feature_names[i]
        for i in topic.argsort()[-10:][::-1]
    ]

    print(f"Topic {topic_idx}: {top_words}")
```

```python
import matplotlib.pyplot as plt

feature_names = vectorizer18.get_feature_names_out()

n_topics = 12

fig, axes = plt.subplots(
    nrows=4,
    ncols=3,
    figsize=(18, 14)
)

axes = axes.flatten()

for idx, topic_id in enumerate(range(n_topics)):

    topic = nmf_model18.components_[topic_id]

    top_indices = topic.argsort()[-10:][::-1]

    words = [
        feature_names[i]
        for i in top_indices
    ]

    weights = topic[top_indices]

    axes[idx].barh(
        words[::-1],
        weights[::-1]
    )

    axes[idx].set_title(
        f"Topic {topic_id}"
    )

plt.suptitle(
    "NMF Topics - World Cup 2018",
    fontsize=16
)

plt.tight_layout()

plt.show()
```
