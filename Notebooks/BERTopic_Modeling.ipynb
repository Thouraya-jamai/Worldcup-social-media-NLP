```python
top_topics = (
    df18["topic"]
    .value_counts()
    .head(21)
)
top_topics
```

```python
#remove -1 (because in BERTopic -1 is usually the outlier topic)
top_topics = top_topics[top_topics.index != -1]
top_topics
```

```python
for topic_id, frequency in top_topics.items():

    print(f"Topic {topic_id}")
    print(f"Frequency: {frequency}")
```

```python
rows = []

for topic_id, frequency in top_topics.items():

    words = [
        word
        for word, score
        in topic_model18.get_topic(topic_id)[:10]
    ]
    rows.append({
        "Topic": topic_id,
        "Frequency": frequency,
        "Words": " | ".join(words)

           
    })

summary_df = pd.DataFrame(rows)

summary_df
```
